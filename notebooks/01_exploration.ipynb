**Question 7** : Expliquer dans une cellule Markdown ce qu'est la lazy evaluation dans Spark. Quelle est la différence entre une
transformation et une action ? Donner 3 exemples de chaque

La lazy evaluation (évaluation paresseuse) signifie que lorsqu'on applique des trasnformations sur un DataFrame, Spark ne les exécute pas immédiatement.
Les traitements sont exécutés uniquement lorsqu'un action demande un résultat.

### Exemple de transformation
Une transformation va créer un nouveau DataFrame à partir de l'existant, sans déclancher immédiatement le calcul. 
- `filter()` : Filtre les lignes
- `distinct()` : retire les lignes identiques
- `groupBy()` : prépare un regroupement pour une agrégation

### Exemple d'action
Une action déclenche l'exécution des transformations définies afin d'obtenir un résultat
- `show()` : Afficher les données
- `count()` : Retourner le nombre de lignes
- `head(5)` : récupèrer les cinq premières lignes

Par exemple :

`df.filter(df.country == "France")`

ça ne déclenche pas immédiatement le traitement. Spark attend qu'une action soit appelée.

`df.filter(df.country == "France").show()`

C'est l'appel à l'action `show()` qui déclenche réellement l'exécution du traitement.

**Question 8** : Ouvrir http://localhost:4040 dans le navigateur. Observer les jobs exécutés. Identifier ce que représentent les
stages et tasks.

Les jobs représentent les actions spark (par exemple je peux voir les count(), ou encore les show() que j'ai pu faire.)

Un job est découpé en plusieurs étapes d'exécution, c'est les stages. Un nouveau stage est générélement crée lorsque Spark doit redistribuer
les données entre les partitions (shuffle)

Une task est une plus petite unité de travail d'un stage. Chaque task traite une partition des données
et plusieurs tasks peuvent être exécutées en parallèle. 

**Question 9** : Lire le fichier orders.csv avec Pandas et mesurer le temps. Puis avec Spark. Comparer. Expliquer quand Spark
est plus pertinent que Pandas.

```python
import pandas as pd
import time

start = time.perf_counter()

df_pandas = pd.read_csv("../data/orders.csv")

end = time.perf_counter()

logger.info(f"Temps Pandas : {end - start:.4f} secondes")

start = time.perf_counter()

df_spark = spark.read.csv("../data/orders.csv", header=True, inferSchema=True)

df_spark.count()

end = time.perf_counter()

logger.info(f"Temps Spark : {end - start:.4f} secondes")
```

Sur ce fichier, Pandas est plus rapide que Spark car le dataset est petit, il ne fait que 830 lignes.

Pandas est plus adapté aux petits volumes de données pouvant être traités en mémoire sur une seule machine.

Spark devient plus pertinent quand le volume de données est important, notamment lorsqu'il est nécessaire de distribuer les traitements sur plusieurs machines et de les exécuter en parallèle.
